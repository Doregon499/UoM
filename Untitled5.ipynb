```python
import time, math, random, os, numpy as np, tensorflow as tf
!pip install fbm
import pandas as pd
from fbm import FBM
from sklearn.metrics import (precision_score, recall_score, f1_score,
                             roc_auc_score, average_precision_score, log_loss)
from tensorflow.keras.utils import to_categorical
```

```python
SEED         = 42
np.random.seed(SEED); tf.random.set_seed(SEED); random.seed(SEED)
BATCH_SIZE   = 128
EPOCHS       = 10
TOTAL_STEPS  = math.ceil(60000 * 0.8 / BATCH_SIZE) * EPOCHS  # 48k train
DEV          = "/GPU:0" if tf.config.list_physical_devices("GPU") else "/CPU:0"
```

```python
(x_train, y_train), _ = tf.keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255.0
x_tr, x_val = x_train[:48000], x_train[48000:]
y_tr, y_val = y_train[:48000], y_train[48000:]
y_tr_cat  = to_categorical(y_tr, 10)
y_val_cat = to_categorical(y_val, 10)

def make_dataset(x, y):
    return (tf.data.Dataset.from_tensor_slices((x, y))
            .shuffle(len(x), seed=SEED)
            .batch(BATCH_SIZE, drop_remainder=True)   # constant shape!
            .prefetch(tf.data.AUTOTUNE))

tr_ds = make_dataset(x_tr,  y_tr_cat)
val_ds = make_dataset(x_val, y_val_cat)
```

```python
def make_model():
    with tf.device(DEV):
        inp  = tf.keras.Input(shape=(28, 28))
        x    = tf.keras.layers.Flatten()(inp)
        x    = tf.keras.layers.Dense(128, activation="relu")(x)
        out  = tf.keras.layers.Dense(10,  activation="softmax")(x)
        return tf.keras.Model(inp, out)
```

```python
def evaluate(model):
    def _pred(ds):
        p, y = [], []
        for xb, yb in ds:
            p.append(model(xb, training=False).numpy())
            y.append(yb.numpy())
        return np.vstack(p), np.vstack(y)

    tr_prob, tr_one = _pred(tr_ds)
    va_prob, va_one = _pred(val_ds)
    tr_lbl, va_lbl  = tr_one.argmax(1), va_one.argmax(1)
    tr_pred, va_pred = tr_prob.argmax(1), va_prob.argmax(1)

    macro = lambda f, a, b: f(a, b, average="macro")
    auc   = lambda y, p: roc_auc_score(y, p, multi_class="ovr", average="macro")

    tr_prec = macro(precision_score, tr_lbl, tr_pred)
    tr_rec  = macro(recall_score,    tr_lbl, tr_pred)
    tr_f1   = macro(f1_score,       tr_lbl, tr_pred)
    tr_auc  = auc(tr_one, tr_prob)
    tr_pr   = average_precision_score(tr_one, tr_prob, average="macro")
    tr_loss = log_loss(tr_one, tr_prob)
    tr_acc  = (tr_pred == tr_lbl).mean()

    va_prec = macro(precision_score, va_lbl, va_pred)
    va_rec  = macro(recall_score,    va_lbl, va_pred)
    va_f1   = macro(f1_score,       va_lbl, va_pred)
    va_auc  = auc(va_one, va_prob)
    va_pr   = average_precision_score(va_one, va_prob, average="macro")
    va_loss = log_loss(va_one, va_prob)
    va_acc  = (va_pred == va_lbl).mean()

    return dict(
        Train_Prec=tr_prec, Train_Rec=tr_rec, Train_F1=tr_f1,
        Train_ROC=tr_auc,   Train_PR=tr_pr,   Train_CE=tr_loss,
        Val_Prec=va_prec,   Val_Rec=va_rec,   Val_F1=va_f1,
        Val_ROC=va_auc,     Val_PR=va_pr,     Val_CE=va_loss,
        Gap=tr_acc - va_acc
    )

```

```python
def run_plain(name, optimizer):
    model = make_model()
    model.compile(optimizer=optimizer,
                  loss='categorical_crossentropy',
                  jit_compile=True)

    # warm-up (one batch → compile)
    model.train_on_batch(*next(iter(tr_ds)))

    # timed fit (compile cost excluded)
    t0 = time.time()
    model.fit(tr_ds, epochs=EPOCHS, verbose=0)
    runtime = time.time() - t0

    metrics = evaluate(model)
    return {
        "Optimizer": name, "Time (s)": round(runtime, 2),
        "Weight Updates": TOTAL_STEPS, **{k.replace('_', ' '): round(v, 4)
        for k, v in metrics.items()}
    }

```

```python
def fbm_path(hurst, sigma):
    fbm = FBM(n=TOTAL_STEPS, hurst=hurst, length=1.0,
              method="daviesharte").fbm()
    return tf.constant(sigma * np.diff(fbm, prepend=0).astype("float32"))
```

```python
def run_sgd_fbm(eta=0.01, mu=0.9, hurst=0.7, sigma=0.01):
    xi = fbm_path(hurst, sigma)
    model = make_model()
    m_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]
    loss_fn = tf.keras.losses.CategoricalCrossentropy()

    @tf.function(jit_compile=True)
    def step(xb, yb, noise):
        with tf.GradientTape() as tape:
            loss = loss_fn(yb, model(xb, training=True))
        grads = tape.gradient(loss, model.trainable_variables)
        for g, m, w in zip(grads, m_buf, model.trainable_variables):
            m.assign(mu * m + g + noise)
            w.assign_sub(eta * m)

    # compile warm-up
    xb, yb = next(iter(tr_ds))
    step(xb, yb, xi[0])

    # timed loop
    t0 = time.time()
    for i, (xb, yb) in enumerate(tr_ds.repeat(EPOCHS)):
        step(xb, yb, xi[i])
    runtime = time.time() - t0

    metrics = evaluate(model)
    return {
        "Optimizer": "SGD+fBm", "Time (s)": round(runtime, 2),
        "Weight Updates": TOTAL_STEPS, **{k.replace('_', ' '): round(v, 4)
        for k, v in metrics.items()}
    }

```

```python
def run_adam_fbm_fast(eta=1e-3, beta1=0.9, beta2=0.999,
                      eps=1e-8, hurst=0.7, sigma=0.01):
    xi = fbm_path(hurst, sigma)
    model = make_model()
    m_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]
    v_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]
    loss_fn = tf.keras.losses.CategoricalCrossentropy()

    @tf.function(jit_compile=True)
    def step(xb, yb, step_idx, noise):
        t = tf.cast(step_idx + 1, tf.float32)
        with tf.GradientTape() as tape:
            loss = loss_fn(yb, model(xb, training=True))
        grads = tape.gradient(loss, model.trainable_variables)

        for g, m, v, w in zip(grads, m_buf, v_buf, model.trainable_variables):
            m.assign(beta1 * m + (1. - beta1) * g + noise)
            v.assign(beta2 * v + (1. - beta2) * tf.square(g))

            m_hat = m / (1. - tf.pow(beta1, t))
            v_hat = v / (1. - tf.pow(beta2, t))
            w.assign_sub(eta * m_hat / (tf.sqrt(v_hat) + eps))

    # warm-up compile
    xb, yb = next(iter(tr_ds)); step(xb, yb, tf.constant(0), xi[0])

    # timed loop
    t0 = time.time()
    for i, (xb, yb) in enumerate(tr_ds.repeat(EPOCHS)):
        step(xb, yb, tf.constant(i), xi[i])
    runtime = time.time() - t0

    met = evaluate(model)
    return {"Optimizer": "Adam+fBm", "Time (s)": round(runtime, 2),
            "Weight Updates": TOTAL_STEPS,
            **{k.replace('_', ' '): round(v, 4) for k, v in met.items()}}

```

```python
def run_nadam_fbm_fast(eta=1e-3, beta1=0.9, beta2=0.999,
                       eps=1e-8, hurst=0.7, sigma=0.01):
    xi = fbm_path(hurst, sigma)
    model = make_model()
    m_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]
    v_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]
    loss_fn = tf.keras.losses.CategoricalCrossentropy()

    @tf.function(jit_compile=True)
    def step(xb, yb, step_idx, noise):
        t = tf.cast(step_idx + 1, tf.float32)
        with tf.GradientTape() as tape:
            loss = loss_fn(yb, model(xb, training=True))
        grads = tape.gradient(loss, model.trainable_variables)

        for g, m, v, w in zip(grads, m_buf, v_buf, model.trainable_variables):
            m.assign(beta1 * m + (1. - beta1) * g + noise)
            v.assign(beta2 * v + (1. - beta2) * tf.square(g))

            m_hat = m / (1. - tf.pow(beta1, t))
            v_hat = v / (1. - tf.pow(beta2, t))
            g_hat = g / (1. - tf.pow(beta1, t))
            nesterov = beta1 * m_hat + (1. - beta1) * g_hat
            w.assign_sub(eta * nesterov / (tf.sqrt(v_hat) + eps))

    xb, yb = next(iter(tr_ds)); step(xb, yb, tf.constant(0), xi[0])

    t0 = time.time()
    for i, (xb, yb) in enumerate(tr_ds.repeat(EPOCHS)):
        step(xb, yb, tf.constant(i), xi[i])
    runtime = time.time() - t0

    met = evaluate(model)
    return {"Optimizer": "Nadam+fBm", "Time (s)": round(runtime, 2),
            "Weight Updates": TOTAL_STEPS,
            **{k.replace('_', ' '): round(v, 4) for k, v in met.items()}}

```

```python
def run_rmsprop_fbm_fast(eta=1e-3, rho=0.9, mu=0.9,
                         eps=1e-8, hurst=0.7, sigma=0.01):
    xi = fbm_path(hurst, sigma)
    model = make_model()
    r_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]      # E[g²]
    v_buf = [tf.Variable(tf.zeros_like(v), trainable=False)
             for v in model.trainable_variables]      # momentum
    loss_fn = tf.keras.losses.CategoricalCrossentropy()

    @tf.function(jit_compile=True)
    def step(xb, yb, noise):
        with tf.GradientTape() as tape:
            loss = loss_fn(yb, model(xb, training=True))
        grads = tape.gradient(loss, model.trainable_variables)

        for g, r, v, w in zip(grads, r_buf, v_buf, model.trainable_variables):
            r.assign(rho * r + (1. - rho) * tf.square(g))
            rms_step = eta * g / (tf.sqrt(r) + eps)
            v.assign(mu * v + rms_step + noise)
            w.assign_sub(v)

    xb, yb = next(iter(tr_ds)); step(xb, yb, xi[0])  # compile warm-up

    t0 = time.time()
    for i, (xb, yb) in enumerate(tr_ds.repeat(EPOCHS)):
        step(xb, yb, xi[i])
    runtime = time.time() - t0

    met = evaluate(model)
    return {"Optimizer": "RMSprop+fBm", "Time (s)": round(runtime, 2),
            "Weight Updates": TOTAL_STEPS,
            **{k.replace('_', ' '): round(v, 4) for k, v in met.items()}}
```

```python
results = []
# plain
results.append(run_plain("SGD",  tf.keras.optimizers.SGD(0.01, 0.9)))
results.append(run_plain("Adam", tf.keras.optimizers.Adam(1e-3)))
results.append(run_plain("Nadam",tf.keras.optimizers.Nadam(1e-3)))
results.append(run_plain("RMSprop",
             tf.keras.optimizers.RMSprop(1e-3, momentum=0.9)))
results.append(run_sgd_fbm())
results.append(run_adam_fbm_fast())
results.append(run_nadam_fbm_fast())
results.append(run_rmsprop_fbm_fast())

df = pd.DataFrame(results)
print(df.to_string(index=False))
```
